## 6.2. Reference-star differential imaging

<div class="alert alert-block alert-warning"> 
    <b>Important: </b>
<a class="anchor" id="war1"></a>

In all post-processing methods of VIP, **Reference-star Differential Imaging** is triggered whenever a 3D reference cube of PSF images is provided with the optional parameter ``cube_ref``, be it for PSF subtraction or forward modeling. In almost all cases, ``cube_ref`` triggers **RDI** only (i.e., the science cube of images is not used to create the PSF model). The only exceptions are the ``pca_annular`` and ``nmf_annular`` functions, where providing a reference cube triggers by default **ARDI** (i.e., using both the reference cube **and** the images of the science cube which match the rotation threshold defined by ``delta_rot``). For RDI-only annular PCA/NMF reductions, one can simply increase the value of ``delta_rot`` to an arbitrarily large value (e.g., 100).
</div>

### 6.2.1. median-RDI

The most straightforward way to leverage a stack of reference PSF images is to consider its median image as PSF model, and subtract it to each image of the science cube. This can be done with the `median_sub` function:

````python
from vip_hci.psfsub import median_sub
med_rdi = median_sub(cube, angs, cube_ref=cube_ref, nproc=nproc)
````

Let's visualize the results, after masking the central 13 px of the image.

````python
from vip_hci.var import mask_circle

plot_frames(mask_circle(med_rdi, 13), grid=True)
````

As can be seen above, this approach is sufficient to detect exoplanet beta Pic b towards the bottom right of the image (after masking the central part of the image)! This is not too bad, considering we only considered as fake reference cube the science images flipped horizontally and vertically.

The default behaviour will subtract the median reference image of the reference stack. However other options can be used. Although the median is more robust, the mean of the reference stack can be requested instead:

````python
mean_rdi = median_sub(cube, angs, cube_ref=cube_ref, collapse_ref='mean', nproc=nproc)
````

Alternatively, the mean/median reference image can be scaled to match in flux each science image before subtraction - this can be useful if the reference image is known to be brighter or fainter than the science image:

````python
med_rdi_sc = median_sub(cube, angs, cube_ref=cube_ref, collapse_ref='sc_median', nproc=nproc)
````

By default when scaling is requested, the scaling factor is calculated from integrating the flux on the whole science and reference images. But an inner and outer radius can be provided, which can be relevant if a coronagraph is used or contamination is present at large radius, respectively. This is done by adding 2 numbers directly at the end of the ``collapse_ref`` string, separated by '-'. For example, if we want to consider an annulus going from 1 to 3 FWHM for this coronagraphic NaCo dataset (FWHM ~ 5pixels), we can set:

````python
med_rdi_sc515 = median_sub(cube, angs, cube_ref=cube_ref, collapse_ref='sc_median5-15', nproc=nproc)
````

Alternatively, median-RDI can be applied in concentric annuli with a scaling factor calculated per annulus. This is controlled with the ``mode`` argument (set to 'fullfr' by default). The width of the concentric annuli is set in pixels using argument `asize`, and the radius of the first annuli is set with ``radius_int``. Let's set both to the FWHM.

````python
med_rdi_ann = median_sub(cube, angs, mode='annular', asize=fwhm_naco, radius_int=fwhm_naco,
                         cube_ref=cube_ref, collapse_ref='sc_median', nproc=nproc)
````

Let's now visualize all the results obtained above, wihtout any numerical mask:

````python
plot_frames((mean_rdi, med_rdi_sc, med_rdi_sc515, med_rdi_ann),
            vmin=-100, vmax=100, # sets the min and max intensity of the colorbar
            pxscale=pxscale_naco, ang_scale=True, # sets the pixel scale and show x/y axis units in arcsec
            ang_ticksep=0.5/pxscale_naco, ang_ticksep_minor=0.1/pxscale_naco) # define the major and minor ticks to be 0.5'' and 0.1'', resp.
````

As expected the results obtained with the different ``collapse_ref`` options are very similar in this case -- since here we used as fake reference cube the flipped science cube. We also notice strong residuals near the center of the image which shows the limitation of assuming a perfect centro-symmetry stellar halo. Nonetheless the signature of beta Pic b can be seen towards the bottom right of the image.

### 6.2.2. PCA-RDI

Using PCA leveraging the RDI strategy is as simple as calling the `pca` function in VIP with the `cube_ref` argument set to your cube of reference images. For example, to run PCA using 42 principal components:

````python
from vip_hci.psfsub import pca

svd_mode = 'lapack'   # you may want to set it to 'arpack' if you have FORTRAN installed -- see speed trick box
pca_rdi_fr = pca(cube, angs, ncomp=42, cube_ref=cube_ref, svd_mode=svd_mode, nproc=nproc)
````

Let's visualize the results, after masking the central 13 px of the image.

````python
from vip_hci.var import mask_circle
plot_frames(mask_circle(pca_rdi_fr, 13), grid=True)
````

We now see Beta Pic b a bit clearer than with the median-subtraction approach.

A lot of options are available with the PCA function, and are also relevant with a reference cube. We refer to [Tutorial 3 (section 3.5)](https://vip.readthedocs.io/en/latest/tutorials/03A_psfsub_ADI.html#3.5.-Principal-Component-Analysis-(PCA)), for all available options. Below we just illustrate the use of a convenient function providing the SNR of a candidate as a function of number of principal components.

Let's first define the approximate coordinates of the companion based on the image above (remember to subtract 1 to the coordinates read in the image since Python uses a zero-based index system):

````python
xy_b = (58, 35)
````

````python
from vip_hci.psfsub import pca_grid
pca_rdi_opt = pca_grid(cube, angs, cube_ref=cube_ref, fwhm=fwhm_naco, range_pcs=(1,51,1),
                       source_xy=xy_b, mode='fullfr', full_output=True, plot=True,
                       svd_mode=svd_mode, nproc=nproc)
````

We see that the number of PCs that maximizes the SNR at the location of the companion is 16. If you remember tutorial 03, the typical behaviour for the flux of the companion (measured directly in the post-processed image) is to decrease with the number of PCs for PCA-ADI. Here we see that it stays relatively constant past the optimal number of PCs.

### 6.2.3. PCA-ARDI

One can leverage *both* ADI and RDI in cases where there is enough field rotation in your science cube and a cube of reference images is available. This is referred to as *ARDI*. PCA-ARDI is most easily done by calling the `pca_annular` function in VIP, with the `cube_ref` argument set to your cube of reference images. The latter is a bit slow, so a trick for a faster usage in full-frame is also provided further below.

<div class="alert alert-block alert-info"> 
    <b>Note: </b>
<a class="anchor" id="note1"></a>

Mind the different **default** behaviours of the `pca` and `pca_annular` functions when provided a `cube_ref` reference cube. `pca` will adopt by default a **RDI**-only strategy (i.e., not use the science images to calculate the principal components), while `pca_annular` will *by default* follow an **ARDI** strategy (i.e., use both the reference images and the science images that meet a certain rotation threshold to calculate the principal components associated to each image of the cube).
</div>

First let's consider PCA-ARDI in concentric annuli with a width set to the FWHM, using 13 principal components. By default, the rotation threshold used to accept science images in the PCA library increases from 0.1 FWHM to 1.0 FWHM from the innermost to the outermost annulus, respectively. This threshold can be adapted with the `delta_rot` parameter.

````python
from vip_hci.psfsub import pca_annular
asize = cube.shape[-1]//2 - 1

pca_ardi_fr = pca_annular(cube, angs, ncomp=13, asize=fwhm_naco, fwhm=fwhm_naco,
                          cube_ref=cube_ref, svd_mode=svd_mode, nproc=nproc)
````

<div class="alert alert-block alert-info"> 
    <b>Note: </b>
<a class="anchor" id="note1"></a>

The `delta_rot` parameter is expressed in FWHM units, and corresponds to the amount of required linear motion (in pixels) at the middle of each concentric annulus. A single value, a tuple of 2 elements, or a list of `n_annuli` values, can be provided for that argument corresponding to: a single threshold in rotation to be used for all annuli, the lower and upper bounds of a linearly increasing rotation threshold, and the exact rotation threshold to be used for each of all concentric annuli, respectively.
</div>

Let's visualize the results, after masking the central 1 FWHM in the image.

````python
plot_frames(mask_circle(pca_ardi_fr, fwhm_naco), 
            pxscale=pxscale_naco, ang_scale=True,
            ang_ticksep=0.5/pxscale_naco, ang_ticksep_minor=0.1/pxscale_naco)
````

We now see Beta Pic b much clearer - with a result more similar to PCA-ADI ([Tutorial 3 - Section 3.5](https://vip.readthedocs.io/en/latest/tutorials/03A_psfsub_ADI.html#3.5.-Principal-Component-Analysis-(PCA))).

Let's consider a few example cases - the most relevant choice for you will depend on your data:

* **ARDI** using all science and reference images in the PCA library, in 2 concentric annuli:

````python
pca_ardi_all_fr = pca_annular(cube, angs, ncomp=13, asize=asize, cube_ref=cube_ref,
                              delta_rot=0, svd_mode=svd_mode, nproc=nproc)
````

* Pure **RDI** in 2 concentric annuli. This is done by setting `delta_rot` to an arbitrarily high value for no science image to meet the criterion for inclusion in the PCA library.

````python
pca_rdi_ann_fr = pca_annular(cube, angs, ncomp=13, asize=asize, cube_ref=cube_ref,
                             delta_rot=100, svd_mode=svd_mode, nproc=nproc)
````

* **ARDI** in a single annulus, with a rotation threshold of 1 FWHM for inclusion of the science images to the PCA library:

````python
asize1 = cube.shape[-1]//2 - 1
pca_ardi_thr1 = pca_annular(cube, angs, ncomp=13, asize=asize1, cube_ref=cube_ref,
                            delta_rot=1, svd_mode=svd_mode, nproc=nproc)
````

* **ARDI** in 4 concentric annuli, with a rotation threshold linearly increasing from 0.2 FWHM for the innermost annulus up to 1 FWHM for the outermost annulus:

````python
asize4 = cube.shape[-1]//8 - 1
pca_ardi_ann4 = pca_annular(cube, angs, ncomp=13, asize=asize4, cube_ref=cube_ref,
                            delta_rot=[0.1, 0.25, 0.5, 1], svd_mode=svd_mode, nproc=nproc)
````

* **ARDI** in full-frame considering all science and reference images (the **fastest** of all available options for ARDI). Here the trick is to redefine the reference cube to be a concatenation of the original science and reference cubes before passing it to the `pca` function:

````python
cube_ref_tmp = np.concatenate((cube, cube_ref))
pca_ardi_full = pca(cube, angs, ncomp=13, cube_ref=cube_ref_tmp, svd_mode=svd_mode, nproc=nproc)
````

<div class="alert alert-block alert-info"> 
    <b>Note: </b>
<a class="anchor" id="note1"></a>

If you run ARDI without any threshold in rotation on the science images, you may get very similar results to either ADI or RDI depending on the relative number of science and reference images in each cube, as the larger image stack will dominate the principal components.
</div>

Let's visualize the results. For this, let's first place all the processed images in a numpy array, and mask the inner 10px of each image: 

````python
pca_ardi_res = np.array((pca_ardi_fr, pca_ardi_all_fr, pca_rdi_ann_fr,
                         pca_ardi_thr1, pca_ardi_ann4, pca_ardi_full))
labels = ('PCA-ARDI (npc=13, default)', 'PCA-ARDI (npc=13, all ref and sci)', 'PCA-RDI (npc=13)',
          r'PCA-ARDI (1 ann, $\delta \theta$=1FWHM)', r'PCA-ARDI (4 ann, $\delta \theta$=0.2-1FWHM)', 'PCA-ARDI (full-frame)')
pca_ardi_res_mask = mask_circle(pca_ardi_res, 12)
````

````python
plot_frames(pca_ardi_res_mask, rows=2, label=labels,
            pxscale=pxscale_naco, ang_scale=True,
            ang_ticksep=0.5/pxscale_naco, ang_ticksep_minor=0.1/pxscale_naco)
````

Remember that the optimal strategy and relative weighting of science vs reference images will depend on the data at hand... Is a lot of field rotation available? Are the reference images well correlated to the science images? Do the radial profile of the reference and science PSF match?

### 6.2.4. Contrast curves

Let's now calculate and compare contrast curves obtained with PCA-ADI, PCA-RDI and PCA-ARDI for this dataset. For simplicity, we'll just consider a single set-up for each case in terms of number of principal components and rotation threshold.

Note that the correct way to do this is to first estimate the parameters of any companion (see [Tutorial 5A](https://vip.readthedocs.io/en/latest/tutorials/05A_fm_planets.html)) and remove the companion from the data. We quickly do this with the `cube_planet_free` function in the next cell, considering the parameters of beta Pic b reported in Absil et al. (2013).

````python
from vip_hci.fm import cube_planet_free

r_b =  0.452/pxscale_naco # Absil et al. (2013)
theta_b = 211.2+90 # Absil et al. (2013)
f_b = 648.2

cube_emp = cube_planet_free([(r_b, theta_b, f_b)], cube, angs, psfn=psfn)
````

Among the parameters of the `contrast_curve` function, one needs to provide `starphot` which corresponds to the (expected) flux of the star in the coronagraphic images. ## 6.3. Strategies to mitigate self- and over-subtraction

**Self-subtraction** arises when circumstellar signals (disk or planet) are present in the library of images used to learn a model PSF for subtraction. When the components (principal components for PCA, NMF components, etc.) contain circumstellar signals and are projected onto the science images to produce a model PSF, circumstellar signals get partly *self-subtracted*. This typically happens with the angular or spectral differential imaging strategies.

**Over-subtraction** arises when components are projected onto science images which contain circumstellar signals, regardless of whether the components contain circumstellar signals or not. The issue arises from a mis-estimation of the projection factors due to the presence of circumstellar signals in the images onto which the components are projected. This effect can happen for any differential strategy, including reference-star differential imaging.

**Self- and over-subtraction can lead to geometric biases in the case of extended disk signals.** This means that the original morphology of extended disk signals gets distorted (Milli et al. 2012; Christiaens et al. 2019; Juillard et al. 2022):


(Figure from [Juillard et al. 2022](https://ui.adsabs.harvard.edu/abs/2022A%26A...668A.125J/abstract))

This tutorial explores strategies available in VIP to mitigate these distortions, such as **PCA-based data imputation** or **iterative PCA**. Algorithms based on **inverse-problem** approaches have also been proposed in the literature, but  will not be explored in this tutorial. We refer the interested reader to [Juillard et al. (2023)](https://ui.adsabs.harvard.edu/abs/2023A%26A...679A..52J/abstract) for a discussion of inverse-problem and iterative approaches.

### 6.3.1. NACO dataset on beta Pic

Although the NACO dataset obtained with the vortex coronagraph on beta Pic does not achieve the required sensitivity to image the debris disk, this does not prevent IPCA to be applied to correct for the effects of self- and over-subtraction on planet b.

#### 6.3.1.1. Iterative PCA-ADI

As a reminder: here's how the PCA-ADI result looks like for the NACO dataset on beta Pic:

````python
from vip_hci.psfsub import pca

fr_pca_ADI = pca(cube=cube, angle_list=angs, ncomp=20,
                 svd_mode=svd_mode, nproc=nproc)
````

````python
plot_frames(fr_pca_ADI, pxscale=pxscale_naco, ang_scale=True,
            ang_ticksep=0.5/pxscale_naco, ang_ticksep_minor=0.1/pxscale_naco)
````

The flux of planet b in this image is significantly lower than its actual flux. Negative side lobes encasing the planet are also visible. These are the consequences of self- and over-subtraction by the post-processing algorithm. It is possible to iteratively correct for those effects. The principle of iterative PCA-ADI is illustrated in the following figure:

The first iteration $k=0$ (blue arrows) corresponds to a regular PCA reduction. At the end of iteration $k=0$, a post-processed image $E_0$ is produced, from which significant signals are estimated. All pixels with a corresponding STIM map value above a certain threshold (as in the illustration) or simply all positive signals in this image are then considered as potential circumstellar signals and subtracted from the stack of images used to calculate the principal components at the next iteration (red arrows). The process continues either for the provided number of iterations, or until the image $E_k$ does not change any more.

In VIP, `ipca` is implemented in the `greedy` subpackage:

````python
from vip_hci.greedy import ipca
````

The call to `ipca` is made in a similar fashion as for `pca`. In the example below, for 20 iterations requested with a fixed number of principal components (e.g. 15), it is simply a matter of setting the `nit` argument to 20:

````python
fr_ipca_ADI = ipca(cube=cube, angle_list=angs, ncomp=15, nit=20,
                   svd_mode=svd_mode, nproc=nproc)
````

<div class="alert alert-block alert-info"> 
    <b>Note: </b>
<a class="anchor" id="note1"></a>

The `GreeDs` implementation of IPCA, described in Juillard et al. (2024), is available [here](https://github.com/Sand-jrd/GreeDS). This is a fast and robust implementation of IPCA in `pytorch`. Provided that you have `pytorch` installed in your environment, this implementation can also be called through the `ipca` function of VIP by setting `mode='Juillard23'`. The interested reader is referred to the [GreeDs package](https://github.com/Sand-jrd/GreeDS) for more details.
</div>

Let's plot the comparison between the PCA and IPCA results:

````python
%matplotlib inline
from vip_hci.var import mask_circle
plot_frames((fr_pca_ADI, fr_ipca_ADI), pxscale=pxscale_naco, ang_scale=True,
            ang_ticksep=0.5/pxscale_naco, ang_ticksep_minor=0.1/pxscale_naco)
````

The negative side lobes which are typical of companions in ADI-processed images have almost entirely disappeared, and IPCA-ADI recovered a significant fraction, if not almost all the flux which was self-subtracted.

The result above is not too bad, but some noise near the center of the image was propagated -- as by default all positive signals are propagated. Let's set a threshold `thr` to only try propagating significant signals.

````python
fr_ipca_ADI_thr = ipca(cube=cube, angle_list=angs, 
                       ncomp=15, nit=20, thr=1,
                       strategy='ADI', svd_mode=svd_mode, nproc=nproc)
````

<div class="alert alert-block alert-info"> 
    <b>Note: </b>
<a class="anchor" id="note1"></a>

The threshold `thr` can either be a threshold in terms of normalized STIM map (see Tuto 4), or an absolute threshold in terms of pixel intensity in the post-processed image. The behaviour is determined by the `thr_mode` argument, which is by default set to `'STIM'`, but can instead be set to `'abs'`.

In the example above, we set the threshold to be 1 without specifying `thr_mode`, which implictly meant that at each iteration, we only considered pixels whose value in the normalized STIM map is larger than 1 as containing circumstellar signal that is subtracted from the images used in the PCA library in the next iteration. 
This works well for compact signals, but may not work well in presence of extended circumstellar signals covering a significant fraction of the field, or when not much rotation is present in the dataset.
</div>

````python
plot_frames((fr_ipca_ADI, fr_ipca_ADI_thr))
````

The threshold better mitigates noise propagation, but the negative side lobes are still visible, which suggests that not all the flux of the planet was recovered. 
One piece of domain knowledge is not yet included in the algorithm: the fact that neighbouring pixels are correlated (due to the convolution by the instrumental response). The most straightforward way to include it is using a Gaussian kernel, to smooth the post-processed image at each iteration. 
The optimal value of the kernel size (`smooth_ker`) to use is not trivial, and depends on how high-pass filtered is the image at the first iteration, which will depend on the amount of rotation available and number of PCs used. Using a kernel size that is too large will blur the image too much, while too fine a kernel may require waiting more iterations until recovering the full extent of extended signals, and can take the risk of propagating circular artefacts. As a rule of thumb, a kernel size between 1 and 2 px (e.g. `smooth_ker = 1`) often seems to be a good compromise regardless of the instrument (as long as the images are Nyquist sampled).

In the next cell, let's also request all outputs to be returned. This will allow us to inspect the cube of post-processed images produced at all iterations:

````python
res_ipca_ADI_thr = ipca(cube=cube, angle_list=angs, 
                        ncomp=15, nit=20, thr=1,
                        strategy='ADI', smooth_ker=1.,
                        svd_mode=svd_mode, nproc=nproc, full_output=True)
````

The cube of iterations is the second output:

````python
fr_ipca_ADI_thr, it_cube_ADI_thr, sig_images, v, residuals_cube_, stim_cube, it_cube_nd = res_ipca_ADI_thr
````

Let's inspect it:

````python
plot_cubes(it_cube_ADI_thr)
````

The cube of iterations shows how circumstellar signals are progressively recovered, while the effects of self- or over-subtraction get corrected. This time the image obtained after 20 iterations does not show any negative side lobe, but instead a nice PSF (with a hint of an Airy ring), which suggests most of the flux was well recovered.

    <b>Note: </b>
<a class="anchor" id="note1"></a>

You may have noticed that despite requesting for 20 iterations in the second to last call to `ipca`, the routine stopped after 13 iterations. Inside the function, a check is made after each new iteration to determine whether the new iteration brought a *significant* change to the estimated circumstellar signals or not. What is deemed significant can be controlled with the `atol` and `rtol` absolute and relative tolerances, respectively, which are provided as input to `np.allclose` when comparing consecutive interations to determine whether the images are the same or not element-wise within a certain tolerance. `atol` and `rtol` are both set by default to 0.01.
````python
fr_ipca_ADI_thr_tol, it_cube_ADI_thr, sig_images, residuals_cube, residuals_cube_, stim_cube, it_cube_nd = res_ipca_ADI_thr_tol
````

Let's compare with the variant considered earlier:

````python
%matplotlib inline
plot_frames((fr_ipca_ADI_thr, fr_ipca_ADI_thr_tol),
            pxscale=pxscale_naco, ang_scale=True,
            ang_ticksep=0.5/pxscale_naco, ang_ticksep_minor=0.1/pxscale_naco)
````

The final image is almost the same, except that a faint signal to the top left was also recovered. Given its position angle and separation, this may be a tentative detection of the edge-on debris disk of beta Pic.

Be mindful that the optimal value for the threshold (if used) will depend on your data, how faint you may expect the disk to be, and its spatial extent (as it will affect the normalized STIM map used for the threshold). The optimal value for the tolerance will also depend on the format of your data, and in particular on your pixel units for `atol`.

#### 6.3.1.2. Iterative PCA-ARDI

The previous section showed that in case you do not possess a library of reference PSF images, you can carry out iterative PCA using your science cube alone, as long as it presents some angular diversity (i.e. IPCA-ADI). However a drawback of IPCA-ADI is that you may not always be able to fully recover the circumstellar signals that are invariant to rotation (although the smoothing kernel can somewhat help).

[Juillard et al. (2024)](https://ui.adsabs.harvard.edu/abs/2024A%26A...688A.185J/abstract) showed that an iterative PCA algorithm leveraging both ADI and RDI together was one of the most robust ways to reliably recover extended circumstellar signals, even outperforming PCA or NMF with data imputation. The principle is similar as IPCA-ADI, except that both a reference and science image stack are used to calculate the principal components, where circumstellar images are progressively removed from the science images at each new iteration (and from the science images used for the projection of the principal components).

In the next few cells, let's explore the functionalities of the `ipca` function of VIP that are compatible with the ARDI strategy. First let's consider a constant number of PCs used throughout the iterations (e.g. 20), with a smoothing kernel and threshold:

````python
from vip_hci.greedy import ipca

res_ipca_ARDI = ipca(cube=cube, angle_list=angs, ncomp=30, nit=20, cube_ref=cube_ref,
                     strategy='ARDI', smooth_ker=1., thr=1, svd_mode=svd_mode, nproc=nproc,
                     full_output=True)
````

````python
fr_ipca_ARDI, it_cube_ARDI, sig_images, residuals_cube, residuals_cube_, stim_cube, it_cube_nd = res_ipca_ARDI
````

````python
%matplotlib inline
plot_frames(mask_circle(fr_ipca_ARDI, fwhm_naco))
````

The above image shows that the negative side lobes which are typical of companions in ADI-processed images have disappeared, and that IPCA-ARDI recovered well the full PSF of the planet (with even a hint of first Airy ring). Let's visualize below how this happened throughout the iterations:

````python
plot_cubes(it_cube_ARDI, pxscale=pxscale_naco, ang_scale=True,
           ang_ticksep=0.5/pxscale_naco, ang_ticksep_minor=0.1/pxscale_naco)
````

#### 6.3.2.1. Iterative roll subtraction on the F187N images

Before considering IPCA, let us consider iterative roll subtraction. The principle is straightforward: for a datacube of images obtained at only 2 roll angles, use directly the mean image obtained at the second angle as model for the images obtained at the first angle, and vice-versa, then mean-combine the 2 derotated residual images. At each new iteration, then consider the positive signals in the post-processed image, and subtract them (after rotation) from the 2 original images used as model PSF images. This is implemented in the `iroll` function of VIP.

Let's consider 1000 iterations:

````python
nit = 1000
````

Either roll subtraction or classical ADI-based algorithms filter extended signals. Besides geometric artefacts, the resulting signals can look narrower than what would be expected from the natural convolution with the PSF of the instrument. To compensate for this filtering effect, and also more quickly recover all circumstellar signals one can convolve the post-processed image at each iteration by a Gaussian kernel through the `smooth_ker_aft` argument.

    <b>Note: </b>
<a class="anchor" id="note1"></a>

The `smooth_ker_aft` argument of `iroll` is identical to the `smooth_ker` argument of `ipca`. In the case of `iroll`, there is also the possibility to low-pass filter the images *before* roll subtraction using the `smooth_ker_bef` argument.
</div>

````python
smooth_ker_aft= 1. # Gaussian kernel for smoothing
````

````python
from vip_hci.greedy import iroll

res_iroll_f187n = iroll(cube=cube_f187, angle_list=angs_f187, 
                        nit=nit, nproc=nproc, smooth_ker_aft=smooth_ker_aft,
                        full_output=True, verbose=False)
````

    <b>Note: </b>
<a class="anchor" id="note1"></a>

The `smooth_ker_aft` argument can also accept the PSF of the instrument. However, this may not be the best choice. While extended signals may get filtered, these are not filtered down to single-pixel width -- which means convolving with the PSF will broaden the signals too much.

#### 6.3.2.2. IPCA-ADI on the F187N images

Let's now consider IPCA instead of IROLL. In this example dataset, the difference between the 2 algorithms is very subtle, since in either cases for each image the other image is used as model. While IROLL directly considers the other image as model, IPCA first normalizes the eigenimage and finds the optimal projection factor (i.e. scale) to be applied onto the other. If there is some variation between the flux of the 2 images -- be it astrophysical or not in origin -- IPCA can lead to slightly better results.

<div class="alert alert-block alert-info"> 
    <b>Note: </b>
<a class="anchor" id="note1"></a>

In general, a larger PCA library is beneficial for a better modeling of the science PSF images, as long as the PSF images in the library are correlated with the science PSF images. Nonetheless,  the extreme stability of the NIRCam PSF causes almost all the variance to be captured in the first principal component even when considering all individual images before collapse into 1 mean-combined image per roll angle (>99.9% of cumulative explained variance!). Furthermore the remaining very small differences between images may not be well reproduced by a linear combination of components (e.g. if associated with detector undersampling or Gibbs artefacts stemming from FFT-based image shifts). Therefore, mean-combining all images obtained at the different dither positions into a single image per roll angle, thereby averaging out the effects of detector undersampling and Gibbs artefacts, leads to better results.
</div>

### 6.3.3. SPHERE/IRDIS star-hopping datasets

In case you possess a library of reference PSF images that are highly correlated with your science PSF images, obtained for example through a star-hopping observing strategy, a few options are possible for you:
- PCA-RDI with data imputation in a mask covering the circumstellar signals ([Ren et al. 2023](https://ui.adsabs.harvard.edu/abs/2023A%26A...679A..18R/abstract));
- iterative PCA using your reference cube (i.e. IPCA-ARDI or IPCA-RDI; [Juillard et al. 2024](https://ui.adsabs.harvard.edu/abs/2024A%26A...688A.185J/abstract));
- if the whole sequence is very stable, you may consider either of the two above options on the full cube of science and reference images, instead of each hop pair.


Both data imputation and IPCA aim to correct for **over-subtraction**. Note that if the reference PSF images are not well correlated, an ARDI approach may enhance the PSF model correlation by incorporating frames from the dataset itself.

In this section, we use the SPHERE/IRDIS dataset taken in star-hopping mode (where the reference images are very well correlated) for the target SAO 206462, a protoplanetary disk exhibiting prominent spirals.

Extended structures, such as disks, are particularly prone to over-subtraction. In fact, when projecting the principal components onto the original cube, the presence of the disk leads to an overestimation of the projection coefficients, resulting in the so-called over-subtraction effect.


#### 6.3.3.1. PCA-RDI with data imputation

PCA-RDI is optimal when masking regions that might contain disk features before projecting the speckle model onto the dataset. This way, we avoid an overestimation of the projection coefficients due to the presence of disk signal.

More precisely, the method proposed by [Ren et al. (2023)](https://ui.adsabs.harvard.edu/abs/2023A%26A...679A..18R/abstract) consists of two masks: the anchor mask and the boat mask.

The boat mask is used to compute the speckle field model (i.e., the principal components). It consists of a mask covering signal within ~IWA of coronagraphic mask, affected by strong central stellar light leakage

The anchor mask targets a region that contains only speckles. This mask is used to compute the projection coefficients. Typically, for SPHERE/IRDIS datasets, a ring of speckles is located at 0.7 arcseconds, while circumstellar signals tend to lie within a smaller radius. 



````python
from vip_hci.var import get_annulus_segments

ones= np.ones((350,350))
coro = 8
cring_out = 110
cring_in = 70

boat_mask = get_annulus_segments(ones,coro,cring_out-coro, mode="mask")[0]
anchor_mask = get_annulus_segments(ones,cring_in,cring_out-cring_in, mode="mask")[0]

plot_frames((anchor_mask, boat_mask),title=["Anchor","Boat"])
plot_frames((cube_ref_sao[0]*anchor_mask, cube_ref_sao[0]*boat_mask),title=["Anchor","Boat"])
````

Let's set to 30 the number of principal components `ncomp` considered for model creation, and let's run the PCA algorithm (feel free to test other values):

````python
from vip_hci.psfsub import pca
ncomp = 30

mask_rdi2 = (anchor_mask, boat_mask)   # can be a tuple of 2 2D bool array

fr_pca_di2 = pca(cube_sao, angs_sao, ncomp=ncomp, cube_ref=cube_ref_sao, mask_rdi=mask_rdi2, svd_mode=svd_mode, nproc=nproc)
````

To highlight the purpose of the method, let's compare this approach with a PCA that only masks the region within the IWA.

````python
fr_pca_sao = pca(cube_sao, angs_sao, ncomp=ncomp, cube_ref=cube_ref_sao, mask_center_px=8, svd_mode=svd_mode, nproc=nproc)
````

````python
plot_frames((fr_pca_di2,fr_pca_sao, fr_pca_di2-fr_pca_sao), title=["Boat/Anchor masks","No masking","difference"],pxscale=pxscale_sphere, ang_scale=True,
            ang_ticksep=0.5/pxscale_sphere, ang_ticksep_minor=0.1/pxscale_sphere, vmax=2200)
````

Note the difference in intensity between the disk estimated with and without the anchor mask. In particular, observe the over-subtraction effect at the center of the disk, which is corrected by the masking method.

#### 6.3.3.2. IPCA-RDI/ARDI

Alternatively to this method, it is possible to use IPCA to reduce these datasets.
The IPCA approach prevents over-subtraction by progressively removing the disk signal from the original dataset and performing the projection on this modified dataset. As a result, the projection coefficients are no longer biased by the disk signal.

Let's apply IPCA-ARDI with 10 iterations and 2 components and compare it to PCA with data imputation.

````python
res_ipca_ARDI = ipca(cube=cube_sao, angle_list=angs_sao, ncomp=10, cube_ref=cube_ref_sao, nit=15, mask_center_px=8, strategy='ARDI', mode='Juillard23', 
                   svd_mode=svd_mode, nproc=nproc, full_output=True)

fr_ipca_ARDI, it_cube_ARDI, sig_images, residuals_cube, residuals_cube_, stim_cube, it_cube_nd = res_ipca_ARDI
````

````python
%matplotlib inline
plot_frames([fr_ipca_ARDI, fr_pca_di2, fr_ipca_ARDI-fr_pca_di2], title=["IPCA-ARDI","PCA with data-imputation","difference"],pxscale=pxscale_sphere, ang_scale=True,
            ang_ticksep=0.5/pxscale_sphere, ang_ticksep_minor=0.1/pxscale_sphere)
````
